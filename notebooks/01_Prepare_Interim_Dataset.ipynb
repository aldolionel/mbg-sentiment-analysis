# 01 - Prepare Interim Dataset

Notebook ini memverifikasi hasil pembuatan interim dataset untuk MBG crawl dataset. Scope penelitian sudah dikoreksi menjadi analisis sentimen publik terhadap Program MBG pada media sosial. Dataset prototype saat ini berasal dari X/Twitter crawl, bukan TikTok-only.

```python
from __future__ import annotations

import json
from pathlib import Path

import pandas as pd

CURRENT_DIR = Path.cwd()
REPO_ROOT = CURRENT_DIR if (CURRENT_DIR / "data").exists() else CURRENT_DIR.parent

INTERIM_PATH = REPO_ROOT / "data/interim/mbg_crawl_interim.csv"
SUMMARY_PATH = REPO_ROOT / "outputs/reports/01_prepare_interim_summary.json"
VERIFICATION_PATH = REPO_ROOT / "outputs/reports/01_prepare_interim_verification.md"
FIGURE_PATHS = [
    REPO_ROOT / "outputs/figures/01_text_length_distribution_clean.png",
    REPO_ROOT / "outputs/figures/01_top_terms_basic_clean.png",
]

summary = json.loads(SUMMARY_PATH.read_text(encoding="utf-8"))
interim_df = pd.read_csv(INTERIM_PATH)

print(f"Interim dataset: {INTERIM_PATH}")
print(f"Shape: {interim_df.shape}")
print(f"Verification report: {VERIFICATION_PATH}")
print("Figures:")
for figure_path in FIGURE_PATHS:
    print(f"- {figure_path} | exists={figure_path.exists()}")
```

## Privacy-Safe Column Check

Notebook ini tidak menampilkan isi `clean_text` agar tidak membuka kembali teks mentah atau teks pengguna secara langsung. Pemeriksaan difokuskan pada kolom, jumlah baris, dan indikator kualitas.

```python
forbidden_columns = {
    "username",
    "user_id",
    "id_str",
    "conversation_id_str",
    "location",
    "image_url",
    "profile_url",
    "screen_name",
    "author",
    "nickname",
    "name",
}
present_forbidden = sorted(forbidden_columns.intersection(interim_df.columns))
quality_summary = {
    "rows": len(interim_df),
    "columns": len(interim_df.columns),
    "empty_clean_text_count": int(interim_df["is_empty_clean_text"].sum()),
    "duplicate_clean_text_count": int(interim_df["is_duplicate_clean_text"].sum()),
    "forbidden_identifier_columns_present": present_forbidden,
}
pd.Series(quality_summary).to_frame("value")
```
